```python
import numpy as np
def generate_matrix(n, prob=0.5):
    size = n+2
    matrix = np.random.choice([0, 1], size=(size, size), p=[1-prob, prob])
    
    # Set edges to 0 , I did this so when i go through the matrix and check the surroundings i can check everything , dont need ifs
    #i dont if this is actually faster 
    matrix[0, :] = 0      # Top edge
    matrix[-1, :] = 0     # Bottom edge
    matrix[:, 0] = 0      # Left edge
    matrix[:, -1] = 0     # Right edge
    
    return matrix
    
def find_clusters(matrix): # find all clusters using Depth-First Search (DFS)
    n = len(matrix)
    visited = set()
    clusters = [] # list of all possible clusters

    directions = [(0,1),(1,0),(0,-1),(-1,0)] # 4　connected neighbors (right, down, left, up)

    # using recursive implementation => maybe recursive algorithms
    def dfs(y,x,current_cluster): 
        if (y,x) in visited or matrix[y][x] == 0:
            return
        visited.add((y,x))
        current_cluster.append((y,x))
        
        # example:if (y,x)=(1,2) => check if the value is 1 at (0,2),(1,1),(2,2),(1,3) 
        for dy,dx in directions:
            if matrix[y+dy][x+dx] == 1: 
                dfs(y+dy,x+dx,current_cluster)

    for y in range(1,n-1): #without edge
        for x in range(1,n-1):
            current_cluster = [] # start a new potential cluster
            dfs(y,x,current_cluster)
            if current_cluster:
                clusters.append(current_cluster)
    return clusters

def find_path(matrix,clusters): #check whether any cluster spans the entire grid
    for cluster in clusters:
        expected = set(i for i in range(1, len(matrix)-1)) #without edge 
        #to collect unique and ordered x and y from the cluster
        x_axis = set()  
        y_axis = set()
        for x, y in cluster:
            if x not in x_axis:
                x_axis.add(x)
            if y not in y_axis:
                y_axis.add(y)
        if x_axis == expected or y_axis == expected:
            return True #'return cluster' if you need the path
    return False
```
